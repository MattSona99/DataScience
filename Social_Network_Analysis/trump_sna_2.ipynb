## Analisi del Network di Donald J. Trump (2018)

Questo codice si concentra sull'Analisi Strutturale di Rete (SNA) di un dataset relazionale che mappa le interconnessioni politiche e familiari immediate di Donald J. Trump nel 2018.

L'_obiettivo_ è trasformare i dati grezzi in un grafo analizzabile per identificare la struttura, la coesione e i ruoli chiave all'interno del suo Ego Network durante quel periodo.

***Dati Sorgente***

I dati, trovati nel GitHub al seguente [_LINK_](https://github.com/melaniewalsh/sample-social-network-datasets.git), sono composti da tre file principali:

- _trump-edges.csv_: File di testo che definisce gli archi del grafo. Ogni riga rappresenta un legame tra due nodi distinti, dove ciascun nodo identifica univocamente un individuo.

- _trump-nodes.csv_ : File che definisce tutti i nodi del dataset, indicando per ciascuno il nome e l'ID (coincidenti).

- _trump-networl.graphml_: File che incorpora le informazioni dei due documenti precedenti in un unico formato, facilmente leggibile dalla libreria NetworkX.

**Prossimi Passi**

Dopo l'importazione e l'assegnazione degli attributi, si procederà con l'analisi descrittiva (Densità, Grado Medio) e la visualizzazione del grafo.

# Analisi Descrittiva della Topologia 

Lo script calcola e stampa le principali metriche topologiche che descrivono la dimensione, la coesione e la connettività del grafo, gestendo in modo condizionale le metriche che richiedono la connettività della rete.

1. _Metriche Base_

**n_nodes** e **n_edges**: Calcolano il numero totale di individui (nodi) e il numero totale di relazioni (archi) nella rete.

**density**: Calcola la densità del grafo, ovvero il rapporto tra gli archi esistenti e gli archi massimi possibili. Questo indica quanto è connessa la rete nel suo complesso.

**connected**: Verifica se il grafo è connesso, cioè se è possibile raggiungere ogni nodo da ogni altro nodo.

2. _Metriche di Distanza (Condizionali)_

Queste metriche sono calcolate solo se il grafo risulta essere connesso (if connected:), poiché non hanno significato in un grafo disconnesso (formato da più componenti isolate).

**diameter**: Calcola il diametro del grafo, ovvero la distanza massima (il percorso più lungo) tra due nodi qualsiasi della rete.

**radius**: Calcola il raggio del grafo, che è la distanza minima tra il centro della rete e il nodo più distante da esso.

**periphery**: Identifica i nodi che si trovano sulla periferia della rete (i nodi la cui massima distanza da qualsiasi altro nodo è pari al diametro).

3. _Misura della Coesione Locale_

**avg_clustering**: Calcola il coefficiente di clustering medio. Questa metrica indica la probabilità media che due vicini di un nodo siano a loro volta collegati tra loro, misurando la tendenza della rete a formare cluster o cricche locali.

```python
G = nx.read_graphml('Trump_data/trumpworld.graphml')

print('Massimo grado di un nodo:', max(dict(G.degree()).values()))
print('Minimo grado di un nodo:', min(dict(G.degree()).values()))
print('Grado medio dei nodi:',sum(dict(G.degree()).values()) / G.number_of_nodes())


n_nodes = G.number_of_nodes()
n_edges = G.number_of_edges()
density = nx.density(G)
connected = nx.is_connected(G)
avg_clustering = nx.average_clustering(G)

if connected:
    radius = nx.radius(G)
    diameter = nx.diameter(G)
    periphery = nx.periphery(G)
else:
    radius = None
    diameter = None
    periphery = None

print(f"Number of nodes: {n_nodes}")
print(f"Number of edges: {n_edges}")
print(f"Density: {density}")
print(f"Connected: {connected}")
    
if connected:
    print(f"Radius: {radius}")
    print(f"Diameter: {diameter}")
    print(f"Periphery: {periphery}")
else:
    print("Radius, diameter and periphery are not defined for disconnected graphs.")
    
print(f"Average clustering coefficient: {avg_clustering}")
```

## Analisi Qualitativa della Ego Network di Donald J. Trump (2018)

Questa sezione si concentra sull'analisi qualitativa della struttura di rete attraverso l'applicazione di diversi algoritmi di layout. L'obiettivo è estrarre caratteristiche visive del grafo che possano essere direttamente ricondotte alla coesione, alla gerarchia e alla natura delle interconnessioni all'interno dell'organizzazione.

Sono stati utilizzati algoritmi di layout basati sulla forza, topologici e casuali, esplorandone le dimensioni 2D e 3D per ottenere una visione completa:

# I. Layout Basati sulla Forza (Force-Directed)

Questi layout ottimizzano la visualizzazione posizionando i nodi vicini se sono fortemente connessi (simulando attrazione) e lontani se non lo sono (simulando repulsione).

_Kamada-Kawai (2D/3D) e Spring Layout (2D/3D)_:

**Obiettivo**: Rivelare la struttura a cluster (comunità) e i nodi centrali (hub).

**Deduzione Organizzativa**: I raggruppamenti densi che emergono (specialmente in 3D) rappresentano individui con forti e rapidi canali di comunicazione. La visualizzazione 3D è cruciale per risolvere le sovrapposizioni e confermare l'esistenza di nodi che fungono da ponti critici (bottleneck).

# II. Layout Topologici e Circolari

Questi layout si basano sulla struttura predefinita per l'analisi.

_Circular Layout (2D/3D)_:

**Obiettivo**: Visualizzare la distribuzione complessiva delle connessioni.

**Deduzione Organizzativa**: Permette una chiara distinzione visiva, quantificando l'omofilia e la necessità di coordinamento tra iruoli.

_Spiral Layout (Solo 2D)_:

**Obiettivo**: Ordinare i nodi in base a una metrica (spesso il grado o la centralità) lungo una spirale, evidenziando il centro della rete.

# III. Layout Avanzati e Casuali

_ARF Layout (Attribute-aware force-directed - Solo 2D)_:

**Obiettivo**: Ottenere un layout che tiene conto esplicitamente dell'attributo durante il posizionamento, tendendo a raggruppare i nodi con lo stesso attributo.

**Deduzione Organizzativa**: Utile per confrontare la struttura di connessione con la struttura di appartenenza. Se il layout ARF assomiglia molto al layout Kamada-Kawai, suggerisce che l'appartenenza al gruppo è il fattore dominante nella formazione degli archi.

_Random Layout (Solo 2D)_:

**Obiettivo**: Fornire una baseline non strutturata.

**Deduzione Organizzativa**: Viene utilizzato come riferimento per dimostrare che i raggruppamenti visibili negli altri layout non sono casuali ma sono proprietà intrinseche della topologia di rete.

**NOTA**: 

    siccome il tempo per generare la totalità dei grafici 3D potrebbe richiedere qualche minuto, si consiglia di eseguire prima le visualizzazioni 2D di ognuno,poi eventualemente approfondire tramite il plot in 3 dimensioni.

```python
####--- 2D Kamada Kawai layout ---###

node_to_find = 'Donald J. Trump'

# Colori definiti
COLOR_TRUMP = '#0425ba'     # Blu (Trump)
COLOR_NEIGHBOR = '#b33b24'  # Rosso-Marrone (Vicini)
COLOR_OTHER = '#faa357'     # Arancione (Altri)
ORGANIZATION_COLOR = '#2883bf'

# Troviamo la stella di Trump
nodi_collegati_a_trump = []
if node_to_find in G:
    nodi_collegati_a_trump = list(G.neighbors(node_to_find))
    
# Inizializza liste
node_colors_list = []
nodes_to_label_visible = [] # Lista dei nodi per cui mostrare il nome

for node in G.nodes():
    if node == node_to_find:
        node_colors_list.append(COLOR_TRUMP)
        nodes_to_label_visible.append(node)
    elif node in nodi_collegati_a_trump:
        node_colors_list.append(COLOR_NEIGHBOR)
        nodes_to_label_visible.append(node) # Includiamo anche i vicini
    elif G.nodes[node]['type'] == 'Organization':
        node_colors_list.append(ORGANIZATION_COLOR)
    else:
        node_colors_list.append(COLOR_OTHER)


pos = nx.kamada_kawai_layout(G)


plt.figure(figsize=(50, 50))
plt.title(f'Il network di {node_to_find}', fontsize=20)



nx.draw(
    G, 
    pos, 
    with_labels=False, # Nessuna etichetta qui
    node_size=1000,
    node_color=node_colors_list, # **CORREZIONE: Usa node_color per la lista di HEX**
    alpha=0.8
)



label_dict = {node: node for node in nodes_to_label_visible}

nx.draw_networkx_labels(
    G, 
    pos, 
    labels=label_dict, 
    font_size=10, 
    font_color='black',
    font_weight='bold'
)

plt.show()
```

```python
####--- 3D Kamada Kawai layout ---###

node_to_find = 'Donald J. Trump'
nodi_collegati_a_trump = []

# Troviamo la stella di strump
if node_to_find in G:
    vicini = list(G.neighbors(node_to_find))
    nodi_collegati_a_trump = vicini

# Calcolo delle posizioni 3D
pos_3d = nx.kamada_kawai_layout(G, dim=3)

# --- 1. Preparazione degli Array di Coordinate e Colori ---

node_xyz = np.array([pos_3d[v] for v in G.nodes()])

# Creazione della lista dei colori: ROSSO per il nodo specifico, BLU per gli altri
# Dobbiamo garantire che l'ordine dei colori corrisponda all'ordine dei nodi in G.nodes()
node_colors_list = []
node_text_list = [] # Per il testo hover di TUTTI i nodi
node_labels_visible = {} # Per le etichette di testo VISIBILI (solo per il nodo specifico)

# Assumiamo che i nodi in G.nodes() siano ordinati o che l'iterazione sia consistente
# Se G.nodes() non è ordinato, potresti avere problemi di allineamento.
# In questo caso, è meglio iterare su G.nodes() per generare sia i colori che i testi.

ORGANIZATION_COLOR = '#2883bf'

for node in G.nodes():
    if node == node_to_find:
        node_colors_list.append('#0425ba') # Rosso vivo per il nodo specifico
        node_labels_visible[node] = node # Vogliamo il nome del nodo come etichetta visibile
        node_text_list.append(f'<b>{node}</b><br>Grado: {G.degree[node]}') # Testo hover evidenziato
    elif node in nodi_collegati_a_trump:
        node_colors_list.append('#b33b24') # Rosso vivo per il nodo specifico
        node_labels_visible[node] = node # Vogliamo il nome del nodo come etichetta visibile
        node_text_list.append(f'<b>{node}</b><br>Grado: {G.degree[node]}') # Testo hover evidenziato
    elif G.nodes[node]['type'] == 'Organization':
        node_colors_list.append(ORGANIZATION_COLOR)
    else:
        node_colors_list.append('#faa357') # Blu vivo per tutti gli altri nodi
        node_text_list.append(f'Nodo: {node}<br>Grado: {G.degree[node]}') # Testo hover standard

# Estratto le coordinate per gli archi (percorsi x, y, z)
edge_x = []
edge_y = []
edge_z = []
for u, v in G.edges():
    x0, y0, z0 = pos_3d[u]
    x1, y1, z1 = pos_3d[v]
    edge_x.extend([x0, x1, None])
    edge_y.extend([y0, y1, None])
    edge_z.extend([z0, z1, None])

# --- 2. Creazione degli Oggetti Trace ---

edge_trace = go.Scatter3d(
    x=edge_x, y=edge_y, z=edge_z,
    line=dict(width=0.5, color='#888'),
    hoverinfo='none',
    mode='lines')

# Trace dei Nodi
node_trace = go.Scatter3d(
    x=node_xyz[:, 0], y=node_xyz[:, 1], z=node_xyz[:, 2],
    text=node_text_list, # Testo hover per tutti i nodi
    mode='markers',
    hoverinfo='text',
    marker=dict(
        showscale=False, # Non serve una scala di colori con soli due colori discreti
        color=node_colors_list, # <<< Colori ROSSO/BLU assegnati direttamente
        size=10,
        line=dict(width=1, color='DarkSlateGrey')
    )
)

# Trace per le etichette di testo VISIBILI (solo per il nodo specifico)
# Questo trace sarà invisibile per gli altri nodi
label_trace = go.Scatter3d(
    x=[pos_3d[node][0] for node in node_labels_visible.keys()],
    y=[pos_3d[node][1] for node in node_labels_visible.keys()],
    z=[pos_3d[node][2] for node in node_labels_visible.keys()],
    mode='text',
    text=[str(node) for node in node_labels_visible.keys()], # Il testo sarà il nome del nodo
    textfont=dict(
        color='#4d0d01', # Stesso colore del nodo
        size=16,
        family='Arial',
    ),
    hoverinfo='none',
    showlegend=False
)


# --- 3. Visualizzazione finale ---

fig = go.Figure(data=[edge_trace, node_trace, label_trace], # Includi il trace delle etichette
                layout=go.Layout(
                    title=f'<br>Il network di Donald J. Trump',
                    showlegend=False,
                    height=1000,
                    margin=dict(l=0, r=0, b=0, t=50),
                    scene=dict(
                        aspectmode='cube',
                        xaxis=dict(showgrid=True, zeroline=True, showticklabels=False),
                        yaxis=dict(showgrid=True, zeroline=True, showticklabels=False),
                        zaxis=dict(showgrid=True, zeroline=True, showticklabels=False))))

fig.show()
```

## Analisi di Centralità: Identificazione dei Nodi Chiave

Questa sezione esegue il calcolo delle quattro metriche di centralità fondamentali sulla rete G e le organizza in un DataFrame per classificare i nodi in base al loro ruolo e alla loro influenza strutturale.

Il codice identifica e stampa i Top 5 e i Bottom 5 nodi per ciascuna metrica.

# Metriche Calcolate

Vengono calcolate le seguenti metriche di centralità per ogni individuo nel network:

**Degree Centrality**: Misura il numero di connessioni dirette di un nodo. Indica chi è più attivo o popolare (chi ha più contatti immediati).

**Closeness Centrality**: Misura la vicinanza di un nodo a tutti gli altri. Indica chi può raggiungere il resto della rete più rapidamente (chi è efficiente nella diffusione).

**Betweenness Centrality**: Misura quante volte un nodo si trova sul percorso più breve tra altre coppie di nodi. Indica chi funge da broker o collo di bottiglia nel flusso di informazioni.

**Eigenvector Centrality**: Misura l'influenza di un nodo in base all'influenza dei suoi vicini. Indica chi è connesso a persone importanti (la qualità dei contatti).

# Interpretazione nel Contesto Politico

L'analisi di queste classifiche permetterà di identificare:

_Gli Hub di Contatto_: Nodi con alta Degree Centrality (chi parla con più persone).

_I Broker_: Nodi con alta Betweenness Centrality (chi collega sottogruppi diversi, essenziale per il coordinamento).

_I Nodi Più Influenzati_: Nodi con alta Eigenvector Centrality (chi è strategicamente posizionato vicino ad altri attori influenti).

```python
degree_centrality = nx.degree_centrality(G)
closeness_centrality = nx.closeness_centrality(G)
betweenness_centrality = nx.betweenness_centrality(G)
eigen_centrality = nx.eigenvector_centrality(G)

centrality_df = pd.DataFrame({
    "Node": list(G.nodes()),
    "Degree Centrality": [degree_centrality[node] for node in G.nodes()],
    "Closeness Centrality": [closeness_centrality[node] for node in G.nodes()],
    "Betweenness Centrality": [betweenness_centrality[node] for node in G.nodes()],
    "Eigen Centrality": [eigen_centrality[node] for node in G.nodes()]
})

sorted_centrality_df = centrality_df.copy()
sorted_centrality_df["Node"] = sorted_centrality_df["Node"].astype(str)

sorted_degree = sorted_centrality_df.sort_values(by=["Degree Centrality"], ascending=False)
sorted_closeness = sorted_centrality_df.sort_values(by=["Closeness Centrality"], ascending=False)
sorted_betweenness = sorted_centrality_df.sort_values(by=["Betweenness Centrality"], ascending=False)
sorted_eigen = sorted_centrality_df.sort_values(by=["Eigen Centrality"], ascending=False)

print("Top 5 Degree Centrality:\n", sorted_degree.head(), "\n")
print("Bottom 5 Degree Centrality:\n", sorted_degree.tail(), "\n")

print("Top 5 Closeness Centrality:\n", sorted_closeness.head(), "\n")
print("Bottom 5 Closeness Centrality:\n", sorted_closeness.tail(), "\n")

print("Top 5 Betweenness Centrality:\n", sorted_betweenness.head(), "\n")
print("Bottom 5 Betweenness Centrality:\n", sorted_betweenness.tail(), "\n")

print("Top 5 Eigen Centrality:\n", sorted_eigen.head(), "\n")
print("Bottom 5 Eigen Centrality:\n", sorted_eigen.tail(), "\n")
```

# Considerazioni sulla Centralità del Grafo

Il seguente grafico mostra quattro diverse visualizzazioni della stessa rete, ciascuna colorata in base a una specifica metrica di centralità (Degree, Closeness, Betweenness ed Eigenvector). Lo scopo è identificare visivamente gli attori chiave in base a diverse definizioni di "importanza" all'interno della rete.

1. Interpretazione Generale del Layout

Il layout dei nodi (Kawai Layout) è fissato in tutte le quattro visualizzazioni. Questo è cruciale perché permette di _confrontare_ direttamente come la posizione strutturale di un nodo si traduce nelle diverse metriche.

- Struttura della Rete: La rete appare relativamente _connessa_, ma con una tendenza alla formazione di _cluster_ (raggruppamenti di nodi vicini), tipica delle reti sociali e organizzative.

2. Analisi Comparativa delle Centralità (Colore)

La colormap 'Plasma' (scala continua) usata per i nodi indica che i nodi con i valori più alti (più centrali) sono gialli/bianchi e quelli con i valori più bassi (più periferici) sono viola scuro.

a. **Degree Centrality** (Grado)

- Osservazione: Si nota un nodo o un piccolo gruppo di nodi che spicca in giallo brillante al centro della rete.

- Deduzione: Questo attore è l'*_Hub di contatto principale_*, con il più alto numero di collegamenti diretti (è l'attore più attivo o popolare).

b. **Closeness Centrality** (Vicinanza)

- Osservazione: L'area centrale e alcuni nodi adiacenti mantengono colori giallo/arancio (alti valori).

- Deduzione: I nodi centrali possono diffondere le informazioni al resto della rete con grande _rapidità_ (sono efficienti dal punto di vista della distanza). C'è una forte correlazione tra Grado e Vicinanza.

c. **Betweenness Centrality** (Intermediazione)

- Osservazione: Il colore giallo/bianco non è necessariamente limitato ai nodi con il grado più alto, ma si estende a nodi che si trovano tra i cluster.

- Deduzione: Questo rivela gli attori che fungono da _Broker_ o _Ponti_. Sono individui che, pur non avendo necessariamente il grado più alto, sono indispensabili per il *flusso di informazioni* tra diverse parti della rete. La loro rimozione frammenterebbe la rete più di quanto farebbe la rimozione di un nodo con alto grado ma locale.

d. **Eigenvector Centrality** (Autovettore/Influenza)

- Osservazione: La colorazione gialla/arancio si concentra sui nodi collegati ai nodi con alta Degree o Betweenness.

- Deduzione: Questi attori sono importanti non per quanto sono connessi, ma secondo a chi sono connessi. L'influenza è elevata per i nodi che fanno parte di una cerchia ristretta di attori già influenti (la qualità dei contatti).

3. **Conclusione Strutturale**

La comparazione visiva suggerisce che:

- La rete ha un chiaro punto focale (alto Grado e Closeness).

- Modularità: La modularità è una metrica che misura la forza della divisione del grafo in moduli. Un valore di modularità alto indica che le connessioni all'interno dei gruppi sono significativamente più dense delle connessioni tra i gruppi.

2. **Mappatura e Visualizzazione**

- Disegno del Grafo: Il grafo viene disegnato utilizzando il layout (pos) precedentemente calcolato (Spring Layout). I nodi sono colorati in base alla loro comunità di appartenenza.

Il fatto che l'algoritmo sia riuscito a dividere chiaramente la rete in diversi gruppi (colori) indica che la rete possiede un'alta modularità. Questo significa che le connessioni all'interno dei gruppi sono molto più dense e frequenti rispetto alle connessioni tra gruppi diversi.

- Identificazione e Ruoli dei Gruppi

Il grafo è visivamente dominato da diversi cluster primari e secondari.

- Cluster Primari (Coesione): Si osservano due o tre grandi raggruppamenti principali (ad esempio, il gruppo viola in alto, il gruppo verde/giallo al centro-sinistra e il gruppo rosso/arancio in basso). Questi grandi gruppi rappresentano le principali fazioni, divisioni organizzative o aree tematiche del network. I nodi al loro interno sono strettamente interconnessi.

- Cluster Secondari/Periferici: Ci sono anche gruppi più piccoli (es. un piccolo gruppo giallo isolato in alto a destra) che potrebbero rappresentare unità periferiche, cellule dormienti o attori esterni collegati solo in minima parte alla struttura centrale.

- Ponti e Broker: Gli individui che si trovano sui bordi di un cluster (ad esempio, un nodo rosso con un arco verso un nodo viola) sono attori cruciali che fungono da "ponti" o broker. La loro rimozione potrebbe isolare intere sezioni della rete.

# Analisi della Coesione e della Struttura Profonda

Questa sezione del codice si focalizza sull'analisi delle sottostrutture coese e sulla gerarchia di densità interna del grafo, utilizzando metriche che vanno oltre la semplice centralità.

1. Rilevamento di Sottostrutture Coese

Il codice calcola due misure fondamentali di coesione:

- **Cricche** (Cliques): Vengono identificate tutte le cricche massimali presenti nel grafo (sottografi in cui ogni nodo è connesso a ogni altro nodo).

    - Risultato (Clique): Restituisce il numero totale di cricche massimali trovate. Nelle reti sociali, le cricche rappresentano i gruppi più coesi e fidati (dove tutti si conoscono).

- **Chiusure Triadiche** (Triadic Closures): Viene calcolato il numero totale di triangoli (o triadi) presenti nel grafo.

    - Risultato (Triadic_closures): Restituisce il numero totale di triadi. Un alto numero di chiusure triadiche suggerisce che la rete è basata sulla fiducia e sulla ridondanza (se A conosce B e B conosce C, è probabile che A conosca C).

2. Analisi del K-Core (Gerarchia di Densità)

Il codice esegue l'analisi del k-core, un metodo per identificare la gerarchia di densità all'interno del grafo.

- Definizione: Il k-core è il sottografo massimo in cui ogni nodo ha almeno k collegamenti all'interno di quel sottografo.

- Processo: Lo script itera da k=1 fino al grado massimo del grafo, identificando i core più densi.

- Risultato (k_core): Viene restituito un dizionario che mostra la dimensione (numero di nodi) di ciascun k-core.

    - Interpretazione Organizzativa: I k-core con k più alto rappresentano il nucleo operativo più interno e denso della rete (la spina dorsale dell'organizzazione), mentre i core con k più basso sono gli anelli esterni e la periferia.

```python
cliques = list(nx.find_cliques(G))

triadic_closures = nx.triangles(G)


k_core_dict = {}
for k in range(1, max(dict(G.degree()).values()) + 1):
    k_core = nx.k_core(G, k=k)
    if k_core.number_of_nodes() > 0:
        k_core_dict[k] = k_core


clique_count = len(cliques)
triadic_closures_count = sum(triadic_closures.values()) // 3  # ogni triangolo viene contato tre volte
k_core_summary = {k: len(core.nodes) for k, core in k_core_dict.items()}

print('Clique: ', clique_count)
print('Triadic_closures: ', triadic_closures_count) 
print('k_core: ', k_core_summary) 
```

# Analisi dei Sottografi Nucleari (Core-Periphery)

Questa sezione identifica le sottostrutture più influenti, coese e resistenti del grafo. L'obiettivo è isolare il nucleo operativo del network e definire la cerchia di contatti immediati dell'attore chiave, Donald J. Trump.

1. Identificazione dei Sottografi Massimi

Il codice estrae le strutture più compatte basate sui risultati di coesione precedentemente calcolati:

- Cricca Massima (Largest cliques): Viene identificata la dimensione della cricca più grande nel grafo (max_clique_size), e l'output mostra i nodi che compongono tutte le cricche che raggiungono quella dimensione massima.

    - Significato: Questo gruppo rappresenta i membri più interconnessi e coesi dell'organizzazione (dove tutti conoscono tutti).

- Core con Densità Massima (Max k e Largest k core):

    - Viene identificato il valore massimo di k (numero di connessioni interne) per cui un k-core esiste (max_k).

    - L'output mostra l'elenco dei nodi che compongono questo nucleo di massima densità (Largest k core).

    - Significato: Questo nucleo rappresenta la spina dorsale operativa del network, il sottografo più resistente e meno incline alla frammentazione.

2. Analisi dell'Ego Network

- Ego Network (ego\_network): Il codice isola e stampa il sottografo incentrato sul nodo Donald J. Trump (ego_node).

    - Definizione: L'Ego Network include il nodo centrale (ego) e tutti i nodi a lui direttamente collegati (i suoi vicini), insieme agli archi tra questi nodi.

    - Significato: Questa visualizzazione definisce la cerchia immediata di influenza e contatto dell'attore chiave.

```python
max_clique_size = max(len(clique) for clique in cliques)
largest_cliques = [clique for clique in cliques if len(clique) == max_clique_size]


max_k = max(k_core_dict.keys())
largest_k_core = k_core_dict[max_k]

ego_node = 'Donald J. Trump' # Sostituisci con l'id di un nodo
ego_network = nx.ego_graph(G, ego_node)  


print("Largest cliques:",largest_cliques)
print("Max k:", max_k)
print("Nodes of the largest k core:",list(largest_k_core.nodes))
print("Ego-network of node:", ego_node, ":", ego_network)
```
